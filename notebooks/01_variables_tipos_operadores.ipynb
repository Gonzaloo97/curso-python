## 2. Variables

Una **variable** es un nombre que apunta a un valor. Piensa en una **etiqueta pegada a una caja**: la caja guarda el dato, la etiqueta te permite referirte a él.

```python
edad = 29
```
Se lee: *"guarda 29 y llámalo `edad`"*. El `=` **no es "igual que"** como en matemáticas: es **asignación** (de derecha a izquierda: primero se calcula lo de la derecha, luego se le pone el nombre de la izquierda).

````python
edad = 29
nombre = "Gonzalo"
print(nombre)
print(edad)

edad = edad + 1     # en matemáticas esto es absurdo; en Python: "calcula edad+1 y guárdalo en edad"
print(edad)
````

### Reglas para nombrar variables
| ✅ Válido | ❌ No válido | Por qué |
|---|---|---|
| `frecuencia_cardiaca` | `frecuencia cardiaca` | no se permiten espacios |
| `paciente2` | `2paciente` | no puede empezar por número |
| `_temporal` | `mi-variable` | `-` se interpreta como resta |
| `Edad` y `edad` | — | ¡son **dos variables distintas**! (Python distingue mayúsculas) |

**Convención en Python** (se llama *snake_case*): todo en minúsculas y palabras separadas por `_`. Ej.: `presion_sistolica`, `numero_de_muestras`. Usa nombres que expliquen qué guardan: `x` no dice nada, `glucosa_mg_dl` lo dice todo.

## 3. Los cuatro tipos básicos

Cada valor en Python tiene un **tipo**, que determina qué puedes hacer con él.

| Tipo | Nombre | Ejemplos | Para qué |
|---|---|---|---|
| `int` | entero | `72`, `-3`, `0` | contar cosas |
| `float` | decimal ("coma flotante") | `36.6`, `0.5`, `-2.0` | medir cosas |
| `str` | cadena de texto (*string*) | `"Ana"`, `"72"`, `""` | texto |
| `bool` | booleano | `True`, `False` | sí / no |

⚠️ En Python el decimal se escribe con **punto**, no con coma: `36.6`, no `36,6`.

La función `type()` te dice el tipo de cualquier cosa:

````python
print(type(72))
print(type(36.6))
print(type("Ana"))
print(type(True))
print(type("72"))   # ¡ojo! con comillas es texto, aunque parezca un número
````

### ¿Por qué importa el tipo? Porque cambia el comportamiento
El mismo símbolo `+` hace cosas distintas según el tipo:

````python
print(5 + 5)        # int + int  -> suma
print("5" + "5")    # str + str  -> pega los textos (se llama "concatenar")
print("ja" * 3)     # str * int  -> repite el texto
````

````python
# Y algunas combinaciones no tienen sentido. Ejecuta esta celda: va a dar ERROR a propósito.
print("Tengo " + 29 + " años")
````

**Lee el error, no te asustes.** La última línea es la importante:
`TypeError: can only concatenate str (not "int") to str` → *"solo puedo pegar texto con texto, no texto con un entero"*.

Leer los mensajes de error es la habilidad nº 1 de un programador. Casi siempre te dicen exactamente qué pasa y en qué línea.

## 4. Convertir entre tipos

Puedes transformar un valor de un tipo a otro con funciones que se llaman igual que el tipo:

````python
print(int("72"))       # texto -> entero
print(float("36.6"))   # texto -> decimal
print(str(29))         # entero -> texto
print(int(3.99))       # decimal -> entero: ¡CORTA los decimales, no redondea!
print(round(3.99))     # si quieres redondear, usa round()
print(float(5))        # entero -> decimal
````

````python
# Así se arregla el error de antes:
print("Tengo " + str(29) + " años")
````

Esto es muy habitual en la vida real: cuando lees datos de un fichero o de lo que escribe un usuario, **siempre llegan como texto** y tienes que convertirlos para operar con ellos.

## 5. Operadores aritméticos

| Operador | Qué hace | Ejemplo | Resultado |
|---|---|---|---|
| `+` | suma | `7 + 2` | `9` |
| `-` | resta | `7 - 2` | `5` |
| `*` | multiplicación | `7 * 2` | `14` |
| `/` | división (siempre da `float`) | `7 / 2` | `3.5` |
| `//` | división entera (cociente) | `7 // 2` | `3` |
| `%` | resto (módulo) | `7 % 2` | `1` |
| `**` | potencia | `7 ** 2` | `49` |

Los tres últimos son los que más chocan al principio. **Analogía**: tienes 7 caramelos y 2 niños. `7 // 2` = cada niño recibe 3. `7 % 2` = te sobra 1.

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(10 / 2)    # fíjate: da 5.0, no 5. La división / siempre devuelve float
print(2 ** 10)
````

**Truco muy usado con `%`**: un número es par si `numero % 2 == 0` (al dividir entre 2 no sobra nada).

**Orden de operaciones**: el mismo que en matemáticas → paréntesis, potencias, multiplicación/división, suma/resta. Ante la duda, **pon paréntesis**: no cuestan nada y hacen el código más claro.

````python
print(2 + 3 * 4)      # 14
print((2 + 3) * 4)    # 20
````

## 6. Comparaciones y booleanos

Los operadores de comparación **hacen una pregunta** y la respuesta es siempre `True` o `False`:

| Operador | Pregunta |
|---|---|
| `==` | ¿son iguales? |
| `!=` | ¿son distintos? |
| `>` / `<` | ¿mayor / menor? |
| `>=` / `<=` | ¿mayor o igual / menor o igual? |

⚠️ **El error más típico del principiante**: `=` asigna, `==` compara. `x = 5` guarda; `x == 5` pregunta.

````python
latidos = 72
print(latidos > 100)
print(latidos == 72)
print(latidos != 72)

taquicardia = latidos > 100     # el resultado de una comparación se puede guardar
print(taquicardia, type(taquicardia))
````

### Combinar condiciones: `and`, `or`, `not`
- `A and B` → `True` solo si **las dos** son ciertas
- `A or B` → `True` si **al menos una** es cierta
- `not A` → le da la vuelta

Python además permite encadenar rangos como en matemáticas: `60 <= latidos <= 100`.

````python
latidos = 72
temperatura = 38.2

print(latidos > 100 or temperatura > 38)     # ¿hay algún signo de alarma?
print(latidos > 100 and temperatura > 38)    # ¿están los dos?
print(60 <= latidos <= 100)                  # ¿está en rango normal?
print(not (60 <= latidos <= 100))            # ¿está FUERA de rango?
````

## 7. f-strings: mostrar resultados con estilo

Pegar textos con `+` y `str()` es incómodo. La forma moderna son las **f-strings**: pon una `f` delante de las comillas y mete variables entre llaves `{}`.

````python
nombre = "Ana"
edad = 34
imc = 25.308641975

print("Paciente: " + nombre + ", edad: " + str(edad))   # forma antigua, engorrosa
print(f"Paciente: {nombre}, edad: {edad}")             # f-string: mucho más limpio
print(f"IMC: {imc}")
print(f"IMC: {imc:.2f}")         # :.2f = muestra 2 decimales
print(f"El año que viene tendrá {edad + 1} años")   # dentro de {} puedes calcular
````

A partir de ahora, **usa f-strings siempre** que quieras mostrar texto con variables. Es lo que verás en todo el código profesional.
