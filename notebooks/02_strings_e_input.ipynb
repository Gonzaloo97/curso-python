````python
# Celda auxiliar de comprobación. No hace falta que la entiendas (aún).
import hashlib

def _h(v):
    if isinstance(v, float):
        v = round(v, 4)
    return hashlib.sha256((type(v).__name__ + repr(v)).encode()).hexdigest()[:16]

def check(nombre, esperado, tipo=None, pista=""):
    g = globals()
    if nombre not in g:
        print(f"❌ La variable `{nombre}` no existe. ¿La has creado y has ejecutado la celda?")
        return
    v = g[nombre]
    if v is None:
        print(f"❌ `{nombre}` sigue valiendo None: todavía no la has rellenado.")
        return
    if tipo is not None and type(v) is not tipo:
        print(f"❌ `{nombre}` debería ser {tipo.__name__}, pero es {type(v).__name__} (vale {v!r})")
        return
    if _h(v) == esperado:
        print(f"✅ {nombre}")
    else:
        print(f"❌ {nombre} = {v!r} no es correcto." + (f"  Pista: {pista}" if pista else ""))

print("Listo. ¡A por ello!")
````

## 1. Qué es un string, `len()` e `in`

Un **string** es una **secuencia de caracteres**: letras, números, espacios, símbolos... Piensa en un tren 🚃🚃🚃 donde cada vagón lleva un carácter. El espacio también ocupa un vagón.

- `len(texto)` → cuántos caracteres tiene (cuántos vagones)
- `"algo" in texto` → `True` si `"algo"` aparece dentro de `texto`
- `"algo" not in texto` → justo lo contrario

````python
diagnostico = "fibrilación auricular"
print(len(diagnostico))          # el espacio cuenta
print("auricular" in diagnostico)
print("Auricular" in diagnostico) # ¡distingue mayúsculas!
print("ventricular" not in diagnostico)
print(len(""))                   # string vacío
````

## 2. Índices: acceder a un carácter

Cada carácter tiene una **posición** (índice). **En Python se empieza a contar en 0**, no en 1. Además, puedes contar desde el final con números negativos:

```
 texto  =   P    Y    T    H    O    N
 índice →   0    1    2    3    4    5
 negativo  -6   -5   -4   -3   -2   -1
```

Se accede con corchetes: `texto[0]` es `"P"`, `texto[-1]` es `"N"` (el último, siempre, sea cual sea la longitud).

Si pides una posición que no existe, Python da `IndexError`.

````python
texto = "PYTHON"
print(texto[0])
print(texto[3])
print(texto[-1])
print(texto[-2])
print(texto[len(texto) - 1])   # otra forma de pedir el último (más larga)
````

````python
# Error a propósito: el índice 6 no existe (van del 0 al 5)
texto = "PYTHON"
print(texto[6])
````

## 3. Slicing: cortar trozos

Con `texto[inicio:fin]` sacas un **trozo** (*slice*). La regla clave: **`inicio` se incluye, `fin` NO**.

Analogía: es como decir "desde la posición 2 **hasta antes de** la 5".

```python
t = "ELECTROCARDIOGRAMA"
t[0:7]     # "ELECTRO"   → posiciones 0,1,2,3,4,5,6
t[:7]      # "ELECTRO"   → si omites inicio, empieza en 0
t[13:]     # "GRAMA"     → si omites fin, llega hasta el final
t[-5:]     # "GRAMA"     → los 5 últimos (¡muy útil!)
t[:]       # copia entera
```

Hay un tercer número opcional, el **paso**: `texto[inicio:fin:paso]`.
```python
t[::2]     # un carácter sí, otro no
t[::-1]    # ¡al revés!
```

Truco para recordarlo: `t[:n]` = los `n` primeros; `t[-n:]` = los `n` últimos. Y `t[a:b]` tiene `b - a` caracteres.

A diferencia de los índices, el slicing **nunca da error** si te pasas: simplemente corta hasta donde haya.

````python
t = "ELECTROCARDIOGRAMA"
print(t[0:7])
print(t[7:13])
print(t[-5:])
print(t[:3])
print(t[::2])
print(t[::-1])
print(t[5:1000])   # te pasas del final y no pasa nada
````

## 4. Concatenar, repetir e inmutabilidad

Ya lo viste: `+` pega textos y `*` los repite.

**Concepto nuevo e importante: los strings son inmutables.** Una vez creado, un string no se puede modificar. No puedes cambiar una letra:
```python
nombre = "Gonzalp"
nombre[6] = "o"   # TypeError: 'str' object does not support item assignment
```
Lo que sí puedes hacer es **construir un string nuevo** con trozos del viejo y guardarlo (en la misma variable si quieres):
```python
nombre = nombre[:6] + "o"
```
Analogía: un string es como un texto impreso. No puedes borrar una letra, pero sí recortar, pegar e imprimir uno nuevo.

````python
nombre = "Gonzalp"
nombre = nombre[:6] + "o"
print(nombre)
print("-" * 20)
print("ja" * 3 + "!")
````

## 5. Métodos de strings

Un **método** es una función que "pertenece" a un tipo de dato y se llama con un punto: `texto.metodo()`.

Analogía: si el string es un objeto, los métodos son los botones que trae. `texto.upper()` = "pulsa el botón *mayúsculas* de este texto".

⚠️ **Clave**: como los strings son inmutables, los métodos **no modifican el original**: **devuelven uno nuevo**. Si no lo guardas, se pierde:
```python
nombre = "ana"
nombre.upper()           # crea "ANA"... y se pierde
print(nombre)            # sigue siendo "ana"
nombre = nombre.upper()  # ✅ así sí
```

### 5.1 Mayúsculas y minúsculas
| Método | Qué hace | `"hOLA mUNDO"` → |
|---|---|---|
| `.upper()` | todo mayúsculas | `"HOLA MUNDO"` |
| `.lower()` | todo minúsculas | `"hola mundo"` |
| `.title()` | Cada Palabra Con Mayúscula | `"Hola Mundo"` |
| `.capitalize()` | Solo la primera letra del texto | `"Hola mundo"` |

````python
t = "hOLA mUNDO"
print(t.upper())
print(t.lower())
print(t.title())
print(t.capitalize())
print(t)          # el original no ha cambiado
````

### 5.2 Quitar espacios: `strip`
Los datos reales vienen sucios: espacios al principio, al final, saltos de línea...
- `.strip()` quita espacios (y saltos de línea) **de los dos extremos**
- `.lstrip()` solo por la izquierda, `.rstrip()` solo por la derecha
- `.strip("x")` quita ese carácter concreto de los extremos

**Encadenar métodos**: como cada método devuelve un string, puedes aplicar otro detrás: `texto.strip().title()`. Se ejecutan de izquierda a derecha.

## 7. f-strings avanzadas

Dentro de las llaves, después de `:`, puedes indicar **cómo** mostrar el valor. Es el "mini-lenguaje de formato":

| Formato | Ejemplo | Resultado | Para qué |
|---|---|---|---|
| `.2f` | `f"{3.14159:.2f}"` | `3.14` | decimales fijos |
| `,` | `f"{47385107:,}"` | `47,385,107` | separador de miles |
| `.1%` | `f"{0.256:.1%}"` | `25.6%` | porcentaje (multiplica por 100 solo) |
| `05d` | `f"{42:05d}"` | `00042` | rellenar con ceros |
| `<10` | `f"{'Ana':<10}"` | `Ana       ` | alinear a la izquierda en 10 huecos |
| `>10` | `f"{'Ana':>10}"` | `       Ana` | alinear a la derecha |
| `^10` | `f"{'Ana':^10}"` | `   Ana    ` | centrar |
| `*^10` | `f"{'Ana':*^10}"` | `***Ana****` | centrar rellenando con `*` |
| `>8.1f` | `f"{98.46:>8.1f}"` | `    98.5` | se combinan: ancho 8, 1 decimal |

Alinear sirve para hacer **tablas bonitas** en la consola.

````python
print(f"{3.14159:.2f}")
print(f"{47385107:,}")
print(f"{0.256:.1%}")
print(f"{42:05d}")
print(f"[{'Ana':<10}]")
print(f"[{'Ana':>10}]")
print(f"[{'Ana':^10}]")
print(f"{'INFORME':=^30}")
print(f"{'Prueba':<12}{'Valor':>8}")
print(f"{'Glucosa':<12}{98.46:>8.1f}")
print(f"{'Colesterol':<12}{185.0:>8.1f}")
````

## 📌 Resumen del módulo
- Un string es una **secuencia**: `len()`, `in`, `not in`.
- **Índices** desde 0; negativos desde el final (`[-1]` = último). Fuera de rango → `IndexError`.
- **Slicing** `[inicio:fin:paso]`: `fin` no se incluye. `[:n]` primeros n, `[-n:]` últimos n, `[::-1]` al revés.
- Los strings son **inmutables**: los métodos devuelven uno nuevo → **guárdalo**.
- Métodos clave: `upper lower title capitalize strip replace count find startswith endswith isdigit isalpha split join`.
- Se pueden **encadenar**: `texto.strip().lower()`.
- Escapes: `\n` `\t` `\\`; raw strings `r"..."` para rutas de Windows.
- f-strings: `:.2f` `:,` `:.1%` `:05d` `:<10` `:>10` `:^10` `:*^10`.
- `input()` **siempre devuelve str**: convierte con `int()` / `float()`.
